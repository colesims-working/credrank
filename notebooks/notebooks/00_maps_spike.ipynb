# Maps Spike - Exploratory Notebook
Quick sanity check that 'search_places' works and results look reasonable.

```python
from credrank.maps_client import search_places
from credrank.models import PlaceSearchParams

params = PlaceSearchParams(
    query="best pizza",
    location=(37.5407, -77.4360),  # Richmond, VA
    radius_m=5000,
    max_results=10,
)

results = search_places(params)
print(f"Got {len(results)} results")
```

```python
import pandas as pd

df = pd.DataFrame([
    {
        "name": r.name,
        "rating": r.rating,
        "reviews": r.user_rating_count,
        "address": r.address,
        "types": ", ".join(r.types[:3]),
    }
    for r in results
])

df.sort_values("rating", ascending=False)
```
